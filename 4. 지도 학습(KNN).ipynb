```python
# 혼동행렬
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
display = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['도미(0)','빙어(1)'])
display.plot(cmap='Blues')
plt.title('혼동행렬')
plt.show()
```
